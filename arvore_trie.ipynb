```python
def inserir_palavra(arvore, palavra):
    atual = arvore # Começa no topo (raiz) da nossa árvore # Também pega todas as letras na arvore  # Referencia a arvore aqui
    
    for letra in palavra: # Pega cada letra da palavra
        
        if letra not in atual: # Se a letra não existir
            atual[letra] = {} # Cria outro dicionario com esta letra
        
        atual = atual[letra] # Descemos para a letra, a fim de verificar se á mais alguma letra que teremos que adicionar
        
    
    atual["*"] = True # Msrcamos com um caracter para saber o fim de uma palavra
```

```python
def buscar_palavra(arvore, palavra):
    atual = arvore # Começa no topo (raiz) da nossa árvore # Também pega todas as letras na arvore  # Referencia a arvore aqui
    
    for letra in palavra: # Pega cada letra da palavra

        if letra not in atual:  # Se a letra não existir
            return False # A palavra não existe
        
        atual = atual[letra] # Se não parar no if, a letra existe, logo, passa para a próxima letra
        
    return "*" in atual # Se existir, retorna o 
```
